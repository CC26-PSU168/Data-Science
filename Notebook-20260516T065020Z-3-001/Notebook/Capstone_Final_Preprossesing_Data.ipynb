# DATA CLEANING

```python
# CLEANING df1

# Buat salinan df1 agar data asli tidak berubah
df1_clean = df1.copy()

# 1. Konversi kolom Date dari string ke datetime
df1_clean['Date'] = pd.to_datetime(df1_clean['Date'])

# 2. Standarisasi kolom Type jadi huruf besar semua
df1_clean['Type'] = df1_clean['Type'].str.strip().str.upper()

# 3. Standarisasi kolom Category jadi Title Case
df1_clean['Category'] = df1_clean['Category'].str.strip().str.title()

# 4. Ganti 'Uncategorized' menjadi 'Lain-lain'
df1_clean['Category'] = df1_clean['Category'].replace('Uncategorized', 'Lain-lain')

# 5. Rename kolom agar seragam
df1_clean = df1_clean.rename(columns={
    'Title'   : 'Description',
    'Type'    : 'Transaction_Type',
    'Account' : 'Account_Name'
})

print("✅ Cleaning df1 selesai!")
print(df1_clean.dtypes)
print()
print(df1_clean['Category'].unique())
print()
print(df1_clean.head())
```

```python
# CLEANING df3
df3_clean = df3.copy()

# 1. Konversi Date ke datetime
df3_clean['Date'] = pd.to_datetime(df3_clean['Date'])

# 2. Standarisasi Transaction_Type dan Category
df3_clean['Transaction_Type'] = df3_clean['Transaction_Type'].str.strip().str.upper()
df3_clean['Category']         = df3_clean['Category'].str.strip().str.title()

# 3. Pastikan nama kolom sudah sesuai
# Cek kalau masih ada kolom lama, rename
if 'Title' in df3_clean.columns:
    df3_clean = df3_clean.rename(columns={'Title': 'Description'})
if 'Type' in df3_clean.columns:
    df3_clean = df3_clean.rename(columns={'Type': 'Transaction_Type'})
if 'Account' in df3_clean.columns:
    df3_clean = df3_clean.rename(columns={'Account': 'Account_Name'})

# 4. Pastikan kolom yang ada sesuai urutan
df3_clean = df3_clean[['Date', 'Description', 'Amount',
                        'Transaction_Type', 'Category', 'Account_Name']]

print("✅ Cleaning df3 selesai!")
print(df3_clean.dtypes)
print()
print(df3_clean['Category'].unique())
print()
print(df3_clean.head())
```

```python
# CLEANING df_persona

df_persona_clean = df_persona.copy()

# 1. Pastikan Date sudah datetime
df_persona_clean['Date'] = pd.to_datetime(df_persona_clean['Date'])

# 2. Standarisasi Transaction_Type dan Category
df_persona_clean['Transaction_Type'] = df_persona_clean['Transaction_Type'].str.strip().str.upper()
df_persona_clean['Category']         = df_persona_clean['Category'].str.strip().str.title()

# 3. Replace kategori tidak konsisten
df_persona_clean['Category'] = df_persona_clean['Category'].replace({
    'Lain-Lain': 'Lain-lain',
    'Uncategorized': 'Lain-lain'
})

# 4. Hapus duplikat
df_persona_clean = df_persona_clean.drop_duplicates().reset_index(drop=True)

# 5. Pastikan kolom sesuai urutan
df_persona_clean = df_persona_clean[['Date','Description','Amount',
                                      'Transaction_Type','Category','Account_Name']]

print("✅ Cleaning df_persona selesai!")
print(f"Sebelum : {df_persona.shape[0]} baris")
print(f"Sesudah : {df_persona_clean.shape[0]} baris")
print()
print("Kategori:", sorted(df_persona_clean['Category'].unique()))
print()
print("Statistik Amount per kategori:")
print(df_persona_clean.groupby('Category')['Amount'].describe()[['min','mean','max']].to_string())
```

# PERBAIKAN DATA

```python
# MERGE FINAL — df1 + df3 + persona 2,3,4

df_combined = pd.concat([df1_clean, df3_clean, df_persona_clean], ignore_index=True)

# Tambah kolom turunan
df_combined['Month']       = df_combined['Date'].dt.month
df_combined['Month_Name']  = df_combined['Date'].dt.strftime('%B')
df_combined['Day_of_Week'] = df_combined['Date'].dt.day_name()

# Bersihkan
df_combined = df_combined.drop_duplicates().reset_index(drop=True)
df_combined['Category'] = df_combined['Category'].replace({
    'Lain-Lain': 'Lain-lain',
    'Uncategorized': 'Lain-lain'
})

# Cek hasil
print("✅ Merge final selesai!")
print(f"df1_clean  : {df1_clean.shape[0]} baris")
print(f"df3_clean  : {df3_clean.shape[0]} baris")
print(f"df_persona : {df_persona.shape[0]} baris")
print(f"TOTAL      : {df_combined.shape[0]} baris")
print()
print("Kategori:", sorted(df_combined['Category'].unique()))
print()
print(df_combined['Amount'].describe())

# Simpan
df_combined.to_csv(path + 'df_combined_clean.csv', index=False)
print(f"\n💾 Tersimpan!")
```
